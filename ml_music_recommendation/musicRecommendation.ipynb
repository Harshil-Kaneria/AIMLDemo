```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn import tree
import joblib

musicData = pd.read_csv('musicDataSet.csv')
model = DecisionTreeClassifier();

musicData
```

```python
inputData = musicData.drop(columns=['genre'])
outputData = musicData['genre']

# model.fit(inputData,outputData);
# predictions = model.predict([ [29,1],[29,0] ])
# predictions

inputDataTrain,inputDataTest,outputDataTrain,outputDataTest = train_test_split(inputData,outputData,test_size=0.2)
model.fit(inputDataTrain,outputDataTrain);

# predictions = model.predict(inputDataTest)
# score = accuracy_score(outputDataTest,predictions)

joblib.dump(model,'music_recommendation.joblib')
tree.export_graphviz(model,out_file='music_recommendation.dot',
                     feature_names=['age','gender'],
                     class_names=sorted(outputData.unique()),
                     label='all',
                     rounded=True,
                     filled=True
                    )
```
